```python
from pydantic import BaseModel
from dataclasses import dataclass
```

```python
@dataclass
class Person(BaseModel):
    name:str
    age:int
    city:str

```

```python
person = Person(name="Anupam",age=22,city="Bangalore")
print(type(person))
```

### Model with Optional Fields

```python
from typing import Optional
```

```python
class Employee(BaseModel):
    id: int
    name:str
    department:str
    salary:Optional[float] = None
```

```python
employee = Employee(id=2322,name="Anupam Mandal",department="CSE")
```

### Pydantic Field Customization

```python
from pydantic import Field
class Market(BaseModel):
    name:str=Field(min_length=2,max_length=10)
    price:float=Field(gt=0,le=100)


market = Market(name="super",price=1)  # attribute has been attached that the value is to be greater than 0
print(market)
```

```python
print(Market.model_json_schema())
```
