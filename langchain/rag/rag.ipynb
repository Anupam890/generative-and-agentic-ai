# Splitting Data into Chuncks

### Recursively split by text characters


 This text splitter is the recommended one for generic text. It is parameterized by a list of characters. It tries to
split on them in order until the chunks are small enough. The default list is ["\n\n", "\n", " ", ""]. This has the
effect of trying to keep all paragraphs (and then sentences, and then words) together as long as possible, as
those would generically seem to be the strongest semantically related pieces of text.

. How the text is split: by list of characters.
. How the chunk size is measured: by number of characters.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)

final_docs = text_splitter.split_documents(node_docs)

final_docs
```

```python
print(final_docs[0])
print(final_docs[1])
```

# Embedding Techniques

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os

google_api_key = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. Split documents into chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
final_docs = text_splitter.split_documents(node_docs)

# 2. Extract raw text strings
texts = [doc.page_content for doc in final_docs]

# 3. Initialize embedding model
embedded_data_1024 = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-2-preview",
    output_dimensionality=1024
)

# 4. Embed list of texts
embed_data = embedded_data_1024.embed_documents(texts)

len(embed_data), len(embed_data[0])   # should be (#chunks, 1024)
```
